check for missing values in the dataset. This is important as missing values can affect the performance of machine learning models and may require imputation or removal.

```python
print("Duplicate data: ",raw_data.duplicated().sum())
print("Canceled Invoices: ",raw_data['InvoiceNo'].astype(str).str.startswith('C').sum())
print("Price less than 0: ", (raw_data['UnitPrice'] < 0).sum())
print("UnitPrice equal to 0: ",(raw_data['UnitPrice'] == 0 ).sum())
print("Null CustomerID: ",raw_data['CustomerID'].isna().sum())
print("Quantities less then 0: ",(raw_data['Quantity'] < 0).sum())
print("Quantities equal to 0: ",(raw_data['Quantity'] == 0 ).sum())
raw_data.isna().sum()
```

```python
data = raw_data.dropna(subset=['CustomerID'])
print(data.shape)
data = data[~data['InvoiceNo'].astype(str).str.startswith('C')]
print(data.shape)
data = data[data['UnitPrice'] > 0]
print(data.shape)
data = data[data['Quantity'] > 0]
print(data.shape)
data = data[~data.duplicated()]
print(data.shape)
```

### Data Cleaning

The dataset was cleaned by removing records with missing `CustomerID`, cancelled invoices (invoice numbers beginning with `C`), non-positive unit prices, non-positive quantities, and exact duplicate rows.

These filters ensure that the retained records represent identifiable customers and valid positive-value purchases. The cleaned dataset will be used to calculate Recency, Frequency, and Monetary (RFM) features.


```python

print("Unique Customers: ",data['CustomerID'].nunique())
print(f"Date range : starting from: {data['InvoiceDate'].min().date()} to {data['InvoiceDate'].max().date()}")
print("Unique Products: ",data['StockCode'].nunique())
print("Unique Country customers: ",data['Country'].nunique())
print("Total Revenue: ",(data['Quantity'] * data['UnitPrice']).sum())

```
